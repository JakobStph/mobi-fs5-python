```python
import numpy as np
import matplotlib.pyplot as plt
```

**[Task 1.1.]()** Load the image:

```python
imgf = plt.imread('data/NIH3T3/im/dna-0.png')
```

**[Task 1.2.]()** Quantify the image into 256 bins:

```python
img8 = (imgf * 255).round().astype(np.uint8)
```

**[Task 1.3.]()** Compute the image histogram:

```python
h = np.zeros(256)
for i in range(0, 255):
    img_bool = img8 == i
    h[i] = img_bool.sum()
```

Show the computed histogram:

```python
plt.figure(figsize=(14,5))
plt.bar(range(256), h, width=1)
```

**[Task 1.4.]()** Implement the method of Otsu:

```python
min_res = np.inf
best_T = 1
for i in range(1, 256):
    class1 = img8[img8 < i]
    class2 = img8[img8 >= i]
    n1 = len(class1)
    n2 = len(class2)
    var1 = np.var(class1)
    var2 = np.var(class2)
    res = n1 * var1 + n2 * var2
    print(res)
    if res < min_res:
        min_res = res
        best_T = i
print(min_res)
print(best_T)
```

**[Task 1.5.]()** Perform Otsu thresholding:

```python
thres = best_T
image_boolean = img8 >= thres
image_binary = image_boolean.astype(int)
plt.figure(figsize=(7.5, 4))
plt.imshow(image_binary, cmap='gray')
plt.colorbar()
```

**[Task 1.6.]()** Compute the Dice coefficient:

```python
ground_truth = plt.imread('data/NIH3T3/gt/0.png')
ground_truth_boolean = ground_truth > 0.5
cardinality_g = ground_truth_boolean.sum()
cardinality_h = image_boolean.sum()
crossover = image_boolean * ground_truth_boolean
cardinality_crossover = crossover.sum()
dice_coefficient = (2 * cardinality_crossover) / (cardinality_g + cardinality_h)
print(dice_coefficient)
```
