**[Task 2.1.]()** Implement Otsu thresholding as a re-usable function:

```python
def otsu_thresholding(imgf):
    img8 = (imgf * 255).round().astype(np.uint8)
    h = np.zeros(256)
    for i in range(0, 255):
        img_bool = img8 == i
        h[i] = img_bool.sum()
    min_res = np.inf
    best_T = 1
    for i in range(1, 256):
        class1 = img8[img8 < i]
        class2 = img8[img8 >= i]
        n1 = len(class1)
        n2 = len(class2)
        var1 = np.var(class1)
        var2 = np.var(class2)
        res = n1 * var1 + n2 * var2
        if res < min_res:
            min_res = res
            best_T = i
    seg = img8 >= best_T
    return seg
```

**[Task 2.2.]()** Implement a re-usable function to compute the Dice coefficient:

```python
def compute_dice(bin1, bin2):
    assert bin1.dtype == bool
    assert bin2.dtype == bool

    cardinality_g = bin1.sum()
    cardinality_h = bin2.sum()
    crossover = bin1 * bin2
    cardinality_crossover = crossover.sum()
    dice = (2 * cardinality_crossover) / (cardinality_g + cardinality_h)

    return dice
```

```python
dice_values = []
for i in [28, 29, 33, 44, 46, 49]:
    image = plt.imread(f'data/NIH3T3/im/dna-{i}.png')
    ground_truth = plt.imread(f'data/NIH3T3/gt/{i}.png')
    ground_truth_boolean = ground_truth > 0.5
    seg = otsu_thresholding(image)
    dice = compute_dice(seg, ground_truth_boolean)
    dice_values.append(dice)
    print(f'Image {i}: Dice = {dice}')

print(np.mean(dice_values))
```
